```python
import pytesseract
from pdf2image import convert_from_path
from PIL import Image

# For Windows users, you might need to specify the path to your tesseract.exe
# pytesseract.pytesseract.tesseract_cmd = r'C:\Program Files\Tesseract-OCR\tesseract.exe'

def extract_myanmar_text_ocr(pdf_path):
    """
    Extracts Myanmar text from a PDF using OCR.
    
    Args:
        pdf_path (str): The file path to the PDF.

    Returns:
        str: The extracted text.
    """
    text = ""
    try:
        # Convert PDF pages to a list of images
        images = convert_from_path(pdf_path)
        
        print(f"Found {len(images)} pages. Starting OCR process...")

        # Process each image (page)
        for i, image in enumerate(images):
            # Use pytesseract to do OCR on the image, specifying the Myanmar language
            # The 'mya' code tells Tesseract to use the Myanmar language model
            page_text = pytesseract.image_to_string(image, lang='mya')
            text += page_text + "\n\n" # Add a separator between pages
            print(f"Processed page {i+1}/{len(images)}")
            print(text)
        print("OCR complete.")
        return text

    except Exception as e:
        return f"An error occurred: {e}"

# --- Main execution ---
pdf_file = "4381"  # Make sure to include the .pdf extension if your file has it
pdf_text = extract_myanmar_text_ocr(pdf_file)

# Save the correctly extracted text to a file with UTF-8 encoding
if "An error occurred" not in pdf_text:
    with open(f"output_ocr_{pdf_file}.txt", "w", encoding="utf-8") as f:
        f.write(pdf_text)
    print("Successfully extracted text and saved to output_ocr.txt")
else:
    print(pdf_text)
```

```python
import pdfplumber

def convert_zawgyi_to_unicode(text):
    """Comprehensive Zawgyi to Unicode conversion"""
    replacements = {
        # Basic characters
        "ေ": "ေ",
        "ြ": "ြ",
        "ွ": "ွ",
        "ှ": "ှ",
        "့": "့",
        "း": "း",
        "်": "်",
        "ါ": "ာ",
        # Combined characters
        "ေက": "က်",
        "ေခ": "ချ",
        "ေစ": "စေ",
        "ေဆ": "ဆေ",
        "ေည": "ည်",
        # Punctuation
        "။": "။",
        "၊": "၊",
        # Common problematic sequences
        "ေ့": "ေ့",
        "ေး": "ေး",
        "ျ": "ျ",
        "ြ": "ြ",
        "ြွ": "ြွ",
        "ြှ": "ြှ"
    }
    
    # First pass - combined characters
    for zg, uni in replacements.items():
        text = text.replace(zg, uni)
    
    # Second pass - handle remaining characters
    myanmar_chars = [chr(c) for c in range(0x1000, 0x109F+1)]
    cleaned_text = []
    for char in text:
        if char in myanmar_chars or char in ' \n\r\t,.။၊':
            cleaned_text.append(char)
    
    return ''.join(cleaned_text)

def process_pdf(pdf_path):
    """Improved PDF text extraction with better layout handling"""
    full_text = []
    with pdfplumber.open(pdf_path) as pdf:
        for page in pdf.pages:
            # Try both normal and layout-based extraction
            text = page.extract_text(x_tolerance=3, y_tolerance=3)
            if not text or len(text.strip()) < 20:  # Fallback if extraction is poor
                text = page.extract_text(layout=True, x_tolerance=2)
            if text:
                full_text.append(text)
    
    raw_text = "\n".join(full_text)
    return convert_zawgyi_to_unicode(raw_text)

# Example usage
filename='4381'
pdf_text = process_pdf(filename)
print("Extracted text:")
print(pdf_text)

# Save to file with proper encoding
with open(f"output_{filename}_2.txt", "w", encoding="utf-8") as f:
    f.write(pdf_text)
```

```python
import pdfplumber

def convert_zawgyi_to_unicode(text):
    """Enhanced Zawgyi to Unicode conversion with more comprehensive replacements"""
    # Ordered replacements (more specific first)
    replacements = [
        ("ေက", "က်"),
        ("ေခ", "ချ"),
        ("ေစ", "စေ"),
        ("ေဆ", "ဆေ"),
        ("ေည", "ည်"),
        ("ေါ", "ော"),
        ("ေး", "ေး"),
        ("ေ့", "ေ့"),
        ("ျ", "ျ"),
        ("ြ", "ြ"),
        ("ွ", "ွ"),
        ("ှ", "ှ"),
        ("့", "့"),
        ("း", "း"),
        ("်", "်"),
        ("ါ", "ာ"),
        ("ေ", "ေ"),
        (" ့", "့"),  # Handle space before diacritic
        (" း", "း"),  # Handle space before diacritic
        ("ြွ", "ြွ"),
        ("ြှ", "ြှ"),
        ("။", "။"),
        ("၊", "၊")
    ]
    
    # Apply replacements in order
    for zg, uni in replacements:
        text = text.replace(zg, uni)
    
    # Clean up any remaining artifacts
    myanmar_range = range(0x1000, 0x109F+1)
    allowed_chars = {' ', '\n', '\r', '\t', ',', '.', '။', '၊'}
    cleaned_text = []
    
    for char in text:
        if ord(char) in myanmar_range or char in allowed_chars:
            cleaned_text.append(char)
        # Handle some common misplacements
        elif char == '်' and cleaned_text:
            cleaned_text[-1] += char  # Combine with previous character
    
    return ''.join(cleaned_text)

def process_pdf(pdf_path):
    """Improved PDF processing with better text extraction"""
    full_text = []
    with pdfplumber.open(pdf_path) as pdf:
        for page in pdf.pages:
            # Try multiple extraction methods
            text = page.extract_text(x_tolerance=2, y_tolerance=2)
            if not text or len(text.strip()) < 10:
                text = page.extract_text(layout=True, x_tolerance=1)
            
            if text:
                # Clean line breaks and spaces
                text = ' '.join(text.split())
                full_text.append(text)
    
    raw_text = "\n".join(full_text)
    return convert_zawgyi_to_unicode(raw_text)

# Example usage
filename = '4381'  # Make sure to include .pdf extension
pdf_text = process_pdf(filename)
print("Extracted text:")
print(pdf_text)

# Save to file with proper encoding
with open(f"output_{filename}_3.txt", "w", encoding="utf-8") as f:
    f.write(pdf_text)
```

```python
def fix_mixed_encoding(text):
    # Sometimes characters get split incorrectly
    fixes = {
        "ေ က": "က်",
        "ြ ွ": "ြွ",
        "ှ း": "ှး"
    }
    for wrong, right in fixes.items():
        text = text.replace(wrong, right)
    return text
```
